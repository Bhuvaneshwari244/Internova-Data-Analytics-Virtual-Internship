## 2. Data Quality Assessment

The purpose of this stage is to identify missing values, duplicate
records, inconsistent data formats, and potential anomalies.

These checks help us understand the quality and reliability
of the original e-commerce dataset before preprocessing.

## A. Standardize column names

```python
# Remove unwanted spaces from column names
df.columns = df.columns.str.strip()

print("Updated column names:")
print(df.columns.tolist())
```

## B. Check missing values

```python
# Calculate missing values
missing_values = df.isnull().sum()

# Calculate missing percentages
missing_percentage = (missing_values / len(df)) * 100

# Create missing value summary
missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage.round(2)
})

missing_summary = missing_summary.sort_values(
    by="Missing Values",
    ascending=False
)

missing_summary
```

## C. Check duplicate records

```python
# Check completely duplicated rows
print("Exact duplicate rows:", df.duplicated().sum())

# Check duplicate IDs
print("Duplicate IDs:", df["ID"].duplicated().sum())

# Check duplicate Order IDs
print("Duplicate Order IDs:", df["Order_ID"].duplicated().sum())
```

## D. Check unique values in categorical columns

```python
categorical_columns = [
    "Customer_Name",
    "Product",
    "Category",
    "Payment_Method",
    "Status"
]

for column in categorical_columns:
    print("\nColumn:", column)
    print("Unique values:", df[column].nunique())
    print(df[column].value_counts(dropna=False))
```

## 3. Detailed Investigation of Data Quality Issues

In this section, we investigate records containing invalid,
missing, inconsistent, and duplicate values.

The objective is to understand the affected records before
applying suitable data-cleaning techniques.

## A. Investigate duplicate IDs

```python
# Display records with duplicate IDs
duplicate_id_records = df[
    df["ID"].duplicated(keep=False)
].sort_values("ID")

duplicate_id_records
```

## B. Investigate duplicate Order IDs

```python
# Display records with duplicate Order IDs
duplicate_order_records = df[
    df["Order_ID"].duplicated(keep=False)
].sort_values("Order_ID")

duplicate_order_records
```

## C. Investigate invalid Quantity values

```python
# Convert Quantity temporarily to numeric
quantity_numeric = pd.to_numeric(
    df["Quantity"],
    errors="coerce"
)

# Identify missing, non-numeric, and negative quantities
invalid_quantity = df[
    quantity_numeric.isna() | (quantity_numeric <= 0)
]

invalid_quantity[
    ["ID", "Order_ID", "Product", "Quantity"]
]
```

## D. Investigate invalid Price values

```python
# Create a temporary numeric version of Price
price_clean = (
    df["Price"]
    .astype("string")
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

price_numeric = pd.to_numeric(
    price_clean,
    errors="coerce"
)

# Identify invalid and negative prices
invalid_price = df[
    price_numeric.isna() | (price_numeric <= 0)
]

invalid_price[
    ["ID", "Order_ID", "Product", "Price"]
]
```

## E. Investigate invalid dates

```python
# Temporarily convert dates
date_parsed = pd.to_datetime(
    df["Order_Date"],
    format="mixed",
    errors="coerce"
)

# Display invalid dates
invalid_dates = df[date_parsed.isna()]

invalid_dates[
    ["ID", "Order_ID", "Order_Date", "Product"]
]
```

## F. Investigate missing categories

```python
# Display records where Category is missing
missing_category_records = df[
    df["Category"].isna()
]

missing_category_records[
    ["ID", "Product", "Category"]
]
```

## G. Investigate missing totals

```python
# Display records with missing Total
missing_total_records = df[
    df["Total"].isna()
]

missing_total_records[
    ["ID", "Product", "Quantity", "Price", "Total"]
]
```

## 4. Additional Data Validation

This section examines duplicate records, product-category
relationships, total amount discrepancies, and unusual dates.

The objective is to establish reliable cleaning rules
without losing potentially meaningful business information.

## A. Display duplicate records clearly

```python
# Display duplicate IDs without confusing row indexes
duplicate_ids = df[
    df["ID"].duplicated(keep=False)
].sort_values("ID")

display(
    duplicate_ids[
        ["ID", "Customer_Name", "Order_ID",
         "Product", "Quantity", "Price", "Total"]
    ].reset_index(drop=True)
)
```

## B. Verify product-category relationships

```python
# Check which categories are associated with each product
product_category_check = (
    df.dropna(subset=["Category"])
      .groupby("Product")["Category"]
      .value_counts()
      .unstack(fill_value=0)
)

display(product_category_check)
```

## C. Check gross amount against existing Total

```python
# Create temporary numeric columns for comparison
quantity_temp = pd.to_numeric(
    df["Quantity"],
    errors="coerce"
)

price_temp = (
    df["Price"]
    .astype("string")
    .str.replace("$", "", regex=False)
    .str.strip()
)

price_temp = price_temp.replace({
    "four hundred": "400"
})

price_temp = pd.to_numeric(
    price_temp,
    errors="coerce"
)

# Calculate expected gross amount
gross_amount = quantity_temp * price_temp

# Compare with the existing Total
total_comparison = df[
    ["ID", "Product", "Quantity", "Price", "Total"]
].copy()

total_comparison["Gross_Amount"] = gross_amount

total_comparison["Difference"] = (
    total_comparison["Gross_Amount"] -
    total_comparison["Total"]
)

display(total_comparison.head(20))
```

## D. Investigate unusual dates

```python
# Parse dates
parsed_dates = pd.to_datetime(
    df["Order_Date"],
    format="mixed",
    errors="coerce"
)

# Display valid dates before 2024
unusual_dates = df[
    parsed_dates.notna() &
    (parsed_dates.dt.year < 2024)
]

display(
    unusual_dates[
        ["ID", "Order_ID", "Order_Date", "Product"]
    ]
)
```
